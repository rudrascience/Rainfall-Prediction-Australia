### Inefficient Features for Predicting Tomorrow's Rainfall (Data Leakage)

When predicting whether it will rain *tomorrow* using data available up to today, some features are inefficient or practically unusable because they rely on measurements taken at the end of the day or represent the target itself. In a real-time forecasting system, we would not have access to these features in time to make a useful prediction:

1. **`RainTomorrow` (or `RainToday` under the renamed schema)**: This represents the target variable we are trying to predict. Using it as a feature would cause direct target leakage.
2. **`3pm` Measurements (e.g., `Temp3pm`, `Humidity3pm`, `Cloud3pm`, `WindSpeed3pm`, `Pressure3pm`)**: If a prediction for tomorrow needs to be made early in the morning or the night before, any metrics recorded late in the afternoon (3 PM) cannot be used in a timely manner.
3. **`MaxTemp`**: The maximum temperature of the day is typically reached in the afternoon, meaning this metric is not fully known until the day is nearly over.
4. **`Rainfall` / `RainToday` (under original schema)**: The total daily rainfall amount is only finalized at the end of the 24-hour observation cycle, making it unavailable for early forecasts.

If we adjust our approach and aim to predict today’s rainfall using historical weather data up to and including yesterday, then we can legitimately utilize all of the available features. This shift would be particularly useful for practical applications, such as deciding whether you will bike to work today.

With this new target, we should update the names of the rain columns accordingly to avoid confusion.

## Extracting a seasonality feature
Now consider the `Date` column. We expect the weather patterns to be seasonal, having different predictablitiy levels in winter and summer for example.  
There may be some variation with `Year` as well, but we'll leave that out for now.
Let's engineer a `Season` feature from `Date` and drop `Date` afterward, since it is most likely less informative than season.
An easy way to do this is to define a function that assigns seasons to given months, then use that function to transform the `Date` column.


### Create a function to map dates to seasons


````python
def date_to_season(date):
    month = date.month
    if (month == 12) or (month == 1) or (month == 2):
        return 'Summer'
    elif (month == 3) or (month == 4) or (month == 5):
        return 'Autumn'
    elif (month == 6) or (month == 7) or (month == 8):
        return 'Winter'
    elif (month == 9) or (month == 10) or (month == 11):
        return 'Spring'
````

# Points to note - 2
What is the true positive rate? There will be a question on this in the assignment that follows.

The True Positive Rate (TPR), or sensitivity, measures the proportion of actual positive cases correctly identified by a machine learning model.
In Python (using scikit-learn), it is calculated as:

TPR=[ True Positives / (True Positives + False Negatives) ]

Based on the confusion matrix generated for the RandomForestClassifier in Exercise 13, we have the following values:

True Positives (TP) = 284 (the number of actual rainy days correctly predicted as rain)
False Negatives (FN) = 254 (the number of actual rainy days predicted as not raining)
The True Positive Rate (TPR), also known as Sensitivity or Recall, is calculated using the formula:

$$\text{TPR} = \frac{\text{TP}}{\text{TP} + \text{FN}}$$$$\text{TPR} = \frac{\text{TP}}{\text{TP} + \text{FN}}$$

$$\text{TPR} = \frac{284}{254 + 284} = \frac{284}{538} \approx 0.52788$$$$\text{TPR} = \frac{284}{254 + 284} = \frac{284}{538} \approx 0.52788$$
